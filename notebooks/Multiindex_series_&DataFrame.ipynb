## MultiIndex Series in Pandas
A MultiIndex (also known as a Hierarchical Index) in pandas allows you to work with data that has multiple levels of indexing. This feature is particularly useful for handling higher-dimensional data in a more natural way

```python
#lets try to create a multiindex series
index_col = [('epl',2022),('epl',2023),('epl',2024),('laliga',2022),('laliga',2023),('laliga',2024)]
series1 = pd.Series([1007,1008,1009,990,995,999],index=index_col)
print(series1)
type(series1)
```

but there is problem as both league and year are dependent on each other, to solve this we have to use multiindex object

```python
#multiindex from tuples
index_col = [('epl',2022),('epl',2023),('epl',2024),('laliga',2022),('laliga',2023),('laliga',2024)]
multiindex = pd.MultiIndex.from_tuples(index_col)
multiindex
```

```python
multiindex.levels
```

```python
multiindex.levels[0]
```

```python
multiindex.levels[1]
```

```python
# multiindex from product
pd.MultiIndex.from_product([['epl','laliga'],[2022,2023,2024]])  #2d list
```

```python
# creating a series with multiindex object
s = pd.Series([1007,1008,1009,990,995,999],index=multiindex)
s
```

```python
#fetching items from such series
s['epl']
```

```python
s['laliga']
```

```python
s[('epl',2023)]
```

A MultiIndex series in pandas appears to be 2-dimensional due to its hierarchical indexing structure, but it's fundamentally a 1-dimensional data structure.

```python
#unstack
#converts multiindex series into dataframe
temp = s.unstack()         #-----------> inner index is converted into columns
temp
```

```python
# stack
# converts dataframe into a multiindex series
temp.stack()
```

```python
type(temp.stack())
```

### what is the reason of using multiindex series or object?

using multiindex object we can represent higher dimensional data using lower object representation, this is helpful if we want to represent 3d,4d or higher dimensional data , we can do that using multiindex dataframe.

- Using multiindex object we can represent any dimensional data using series and dataframe

```python
 # multiindex dataframe from column perspective i.e multiple index in columns
league_df2 = pd.DataFrame([
    [301,999,305,1005],
    [301,999,305,1005],
    [301,999,305,1005]],
             index = [2022,2023,2024],
             columns = pd.MultiIndex.from_product([['epl','laliga'],['no.of.players','no.of.goals']])
)
league_df2
```

```python
#fetching item from this dataframe
league_df2['epl']
```

```python
league_df2['laliga']
```

```python
league_df2['laliga']['no.of.goals']
```

```python
# multiindex on both rows and columns
# Creating the MultiIndex for rows
row_index = pd.MultiIndex.from_product([[2022, 2023, 2024], ['A', 'B', 'C']], names=['Year', 'Type'])

# Creating the DataFrame with MultiIndex for both rows and columns
league_df3 = pd.DataFrame([
    [301, 999, 305, 1005],
    [301, 999, 305, 1005],
    [301, 999, 305, 1005],
    [301, 999, 305, 1005],
    [301, 999, 305, 1005],
    [301, 999, 305, 1005],
    [301, 999, 305, 1005],
    [301, 999, 305, 1005],
    [301, 999, 305, 1005]],
    index=row_index,
    columns=pd.MultiIndex.from_product([['epl', 'laliga'], ['no.of.players', 'no.of.goals']])
)
league_df3
```

in this dataframe if i want to fetch a particular item i required 4 information(year,type,epl/laliga,no.of.players/no.of.goals) so this data is 4d which is represented as dataframe using multiindex object.
