26. How do you pivot a DataFrame?

- before answering this question first we have to understand pivot and melt along with wide and long data format

### wide format

Wide format is where we have a single row for every data point with multiple columns to hold the values of various attributes

### Long format

long format is where , for each data point we have as many rows as the number of attributes and each row contains the value of a particular attribute for a given data point

![Long vs Wide Data](../images/widevslong.png)

- pivot ---> converts long to wide
- melt -----> converts wide to long

```python
#now lets answer the question,for that lets create a simple dataframe
data = {
    'Date': ['2024-02-01', '2024-02-01', '2024-02-02', '2024-02-02'],
    'City': ['New York', 'Los Angeles', 'New York', 'Los Angeles'],
    'Temperature': [30, 25, 32, 26]
}

df_long = pd.DataFrame(data)
df_long
```

```python
#lets perform pivot on it 
df_long.pivot(index = 'City', columns = 'Date',values = 'Temperature')
#index : What you want as row labels
#columns : What you want as column labels
#values : What you want inside cells
```

in this way we can convert long to wide format using pivot,in this case converting date/time series data into wide format is not a good practice if we have many dates , this is just demonstrate the use of pivot function

29. How do you join two DataFrames using different types of joins?

There are 4 types of joins.
- inner join (default)
- outer join
- left join
- right join

![joins in pandas](../images/joins.jpg)

```python
# first lets create a two datafram as :
customers = pd.DataFrame({
    'customer_id': [101, 102, 103, 104, 105],
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'city': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
    'age': [25, 30, 35, 40, 28]
})

# Creating Orders DataFrame
orders = pd.DataFrame({
    'order_id': [1, 2, 3, 4, 5],
    'customer_id': [101, 103, 104, 106, 102],
    'product': ['Laptop', 'Phone', 'Tablet', 'Monitor', 'Keyboard'],
    'price': [1200, 800, 300, 200, 100]
})
```

```python
customers
```

```python
orders
```

```python
#inner join ----> Get customers who have placed an order. like an intersection
customers.merge(orders,how='inner',on = ['customer_id'])
```

```python
#outer join -----> like an union
#Get all customers and orders, including those without matches.
customers.merge(orders,how='outer',on = ['customer_id'])
```

```python
#left join 
#Get all customers, along with their orders (if any).
customers.merge(orders,how='left',on = ['customer_id'])
```

```python
#right join
#Get all orders, including customers (even if customer details are missing).
customers.merge(orders,how='right',on = ['customer_id'])
```

```python
#cross join
#Get all possible combinations of customers and orders.
customers.merge(orders,how='cross')
```

36. How do you resample a time series DataFrame?

Resampling a time series DataFrame means changing the frequency of the time index by either aggregating or interpolating data points

```python
# Downsampling , Aggregates data to a lower frequency (e.g., from seconds to minutes, or days to months).
# Create a sample time series DataFrame
data = {'Date': pd.date_range(start='2024-01-01', periods=6, freq='D'), 
        'Value': [10, 15, 20, 25, 30, 35]}

df = pd.DataFrame(data).set_index('Date')
```

```python
# Downsample: Convert daily data to monthly sum
df_monthly = df.resample('M').sum()
print(df_monthly)
```

```python
# Upsample: Convert daily data to hourly and forward fill missing values
df_hourly = df.resample('H').ffill()
print(df_hourly.head(10))
```

64. How do you downcast numerical columns to reduce memory usage?

✅ 1. Why Downcast?

Large datasets often use int64 or float64, which consume a lot of memory.
Downcasting reduces memory usage while keeping the same values

```python
# Create a DataFrame with large number types
df = pd.DataFrame({
    'int_col': [1, 200, 3000, 40000],  # int64
    'float_col': [1.5, 2.5, 3.5, 4.5]  # float64
})

# Downcast integers and floats
df['int_col'] = pd.to_numeric(df['int_col'], downcast='integer')
df['float_col'] = pd.to_numeric(df['float_col'], downcast='float')

print(df.dtypes)
```

67. How do you speed up DataFrame operations using vectorization?

Vectorization in Pandas means applying operations on entire columns (or rows) at once, instead of looping through individual elements. This is way faster because Pandas uses NumPy’s optimized C code under the hood.

```python

import time

df = pd.DataFrame({'A': np.random.randint(1, 100, 1_000_000)})

# Using a loop (slow)
start = time.time()
df['B'] = [x * 2 for x in df['A']]
print("Loop Time:", time.time() - start)
```

```python
# Vectorized operation (fast)
start = time.time()
df['B'] = df['A'] * 2
print("Vectorization Time:", time.time() - start)
```

✔ Vectorized approach is often 50-100x faster than loops!

```python
# lets first import dataset again as df
df = pd.read_csv('../data/survey_results_public_2019.csv')
df.info()
```

81. How do you find the most common programming language in the dataset?

```python
df['LanguageWorkedWith'].str.split(';').explode().value_counts().head(10)
```

82. How do you calculate the average salary for each programming language?

```python
df_lang_salary = df[['LanguageWorkedWith', 'ConvertedComp']].dropna()
df_lang_salary = df_lang_salary.assign(Language=df_lang_salary['LanguageWorkedWith'].str.split(';'))
df_lang_salary = df_lang_salary.explode('Language')
df_lang_salary.groupby('Language')['ConvertedComp'].mean().sort_values(ascending=False)
```

85. Most used programming languages together

```python
from itertools import combinations
from collections import Counter

# Split languages and create combinations
pairs = df['LanguageWorkedWith'].dropna().apply(lambda x: list(combinations(x.split(';'), 2)))
pairs = [pair for sublist in pairs for pair in sublist]

# Count most common pairs
Counter(pairs).most_common(10)
```

86. Trends in developers' years of experience

```python
df['YearsCode'].replace({'Less than 1 year': '0', 'More than 50 years': '50'}, inplace=True)
df['YearsCode'] = pd.to_numeric(df['YearsCode'], errors='coerce')
df['YearsCode'].hist(bins=20)
```

88. Salary difference between remote and non-remote workers

89. Filtering out extreme salary values using quantiles

```python
df_filtered = df[df['ConvertedComp'].between(df['ConvertedComp'].quantile(0.05), df['ConvertedComp'].quantile(0.95))]
df_filtered['ConvertedComp'].describe()
```

90. Gender distribution in the dataset

91. Checking if higher experience leads to higher salary

```python
import matplotlib.pyplot as plt

df_plot = df[['YearsCode', 'ConvertedComp']].dropna()
df_plot = df_plot[df_plot['ConvertedComp'] < df_plot['ConvertedComp'].quantile(0.95)]  # Removing outliers
df_plot.groupby('YearsCode')['ConvertedComp'].mean().plot(kind='line', figsize=(10, 5))
plt.xlabel('Years of Experience')
plt.ylabel('Average Salary')
plt.title('Years of Experience vs. Salary')
plt.show()
```

92. Identifying outliers in salary data using IQR

```python
Q1 = df['ConvertedComp'].quantile(0.25)
Q3 = df['ConvertedComp'].quantile(0.75)
IQR = Q3 - Q1
outliers = df[(df['ConvertedComp'] < (Q1 - 1.5 * IQR)) | (df['ConvertedComp'] > (Q3 + 1.5 * IQR))]
outliers.shape
```
