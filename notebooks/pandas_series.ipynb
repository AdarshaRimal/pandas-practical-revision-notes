# Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding any type of data

```python
import numpy as np
import pandas as pd
```

## Series from list

```python
#string
country = ['Nepal','India','China','USA']
pd.Series(country)  #Series is class in pandas
```

```python
#integers
goals = [1,2,3,1,0,0,1,2,3]
pd.Series(goals)
```

in series there are two items(one is index and another is value(that we provide))

```python
#custom index
players = ['ronaldo','messi','lewandowski','benzema','neymar']
goals = [140,129,103,90,43]
pd.Series(goals,index = players)
```

```python
#setting a name
pd.Series(goals,index = players,name = 'Football players ucl goals')
```

## Series from dictinary

```python
goals = {
    'ronaldo':140,
    'messi':129,
    'lewandowski':103,
    'benzema':90,
    'neymar':43
}
```

```python
goals_series = pd.Series(goals,name="Top 5 UCL Scorer")
goals_series
```

## Series Attributes

## series using read_csv

read_csv is one of the most used and important function to load a csv format data

```python
#with one column
type(pd.read_csv('../data/subs.csv'))
```

!!! when pd.read_csv loads and stores data it stores in the format of DataFrame not series !!!

we can convert dataframe into series using squeeze() method

```python
data = pd.read_csv('../data/subs.csv')
data = data.squeeze()
print(data)
type(data)
```

```python
# with 2 columns
vk = pd.read_csv('../data/kohli_ipl.csv')
vk
```

```python
vk = pd.read_csv('../data/kohli_ipl.csv',index_col='match_no')
vk = vk.squeeze()
vk
```
